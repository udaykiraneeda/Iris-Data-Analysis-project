```python
import pandas as pd
```

```python
#=========================================
# Load the dataset
#=========================================
df = pd.read_csv('1) iris.csv')
df
```

```python
print("=" * 60)
print("STEP 1: INITIAL DATA OVERVIEW")
print("=" * 60)
print(f"Shape of dataset: {df.shape}")
print("\nFirst 5 rows:")
print(df.head())
```

```python
print("\nData Types")
print(df.dtypes)
```

```python
#=======================================================
#  Identify and handle missing values
#=======================================================
print("\n" + "=" * 60)
print("STEP 2: MISSING VALUES")
print("=" * 60)
```

```python
missing_counts = df.isnull().sum()
print("Missing values per column:")
print(missing_counts)
```

```python
if missing_counts.sum() > 0:
    # Numeric columns -> impute with column median (robust to outliers)
    numeric_cols = df.select_dtypes(include="number").columns
    for col in numeric_cols:
        if df[col].isnull().sum() > 0:
            median_val = df[col].median()
            df[col] = df[col].fillna(median_val)
            print(f"Filled missing values in '{col}' with median = {median_val}")
```

```python
categorical_cols = df.select_dtypes(include="object").columns
missing_cat = df[categorical_cols].isnull().sum().sum()
if missing_cat > 0:
    for col in categorical_cols:
        if df[col].isnull().sum() > 0:
            mode_val = df[col].mode()[0]
            df[col] = df[col].fillna(mode_val)
            print(f"Filled missing values in '{col}' with mode = {mode_val}")
else:
    print("No missing values found in categorical columns.")
```

```python
#================================================
# Remove duplicate rows
#================================================
print("\n" + "=" * 60)
print("STEP 3: DUPLICATE ROWS")
print("=" * 60)
```

```python
num_duplicates = df.duplicated().sum()
print(f"Number of duplicate rows found: {num_duplicates}")
```

```python
df = df.drop_duplicates().reset_index(drop=True)
print(f"Shape after removing duplicates: {df.shape}")
```

```python
# =====================================================
# Standardize inconsistent data formats
# =====================================================
print("\n" + "=" * 60)
print("STEP 4: STANDARDIZING FORMATS")
print("=" * 60)
```

```python
# Standardize text/categorical columns: strip whitespace, lowercase
categorical_cols = df.select_dtypes(include="object").columns
for col in categorical_cols:
    before_unique = df[col].unique()
    df[col] = df[col].astype(str).str.strip().str.lower()
    after_unique = df[col].unique()
    print(f"Column '{col}' unique values BEFORE: {before_unique}")
    print(f"Column '{col}' unique values AFTER:  {after_unique}")
```

```python
# Ensure numeric columns are proper numeric dtype (coerce errors to NaN, then re-check)
numeric_cols = df.select_dtypes(include="number").columns
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
# Round numeric measurements to a consistent number of decimal places (1 dp, as in original data)
df[numeric_cols] = df[numeric_cols].round(1)
```

```python
# ================================================
#  Final check & save cleaned dataset
# ================================================
print("\n" + "=" * 60)
print("STEP 5: FINAL CLEANED DATASET")
print("=" * 60)
```

```python
print(f"Final shape: {df.shape}")
```
